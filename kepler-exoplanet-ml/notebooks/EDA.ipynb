```python
import os
import numpy as np
import pandas as pd

# 1. Load data dynamically from project directory
script_dir = os.path.dirname(os.path.abspath(__file__)) if '__file__' in globals() else '.'
data_path = os.path.normpath(os.path.join(script_dir, "..", "data", "cumulative_2020.12.30_14.14.11.csv"))

df = pd.read_csv(data_path, comment="#")

# 2. Drop 100% empty columns and non-predictive identifiers
cols_to_drop = [
    "koi_teq_err1",
    "koi_teq_err2",
    "kepid",
    "kepoi_name",
    "kepler_name",
    "koi_pdisposition",
    "koi_tce_delivname",
]
df_clean = df.drop(columns=[c for c in cols_to_drop if c in df.columns])

# 3. Encode Target Variable (CONFIRMED/CANDIDATE -> 1, FALSE POSITIVE -> 0)
df_clean["target"] = df_clean["koi_disposition"].map(
    {"CONFIRMED": 1, "CANDIDATE": 1, "FALSE POSITIVE": 0}
)
df_clean = df_clean.drop(columns=["koi_disposition"])

# 4. Fill numerical missing values with Median
num_cols = df_clean.select_dtypes(include=[np.number]).columns
df_clean[num_cols] = df_clean[num_cols].fillna(df_clean[num_cols].median())

print(
    f"Cleaned Dataset Shape: {df_clean.shape} | Missing Values: {df_clean.isnull().sum().sum()}"
)
```

```python
output_path = os.path.normpath(os.path.join(script_dir, "..", "data", "kepler_cleaned.csv"))
df_clean.to_csv(output_path, index=False)

print(f"Saved cleaned dataset to: {output_path}")
```
