```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load cleaned data
script_dir = os.path.dirname(os.path.abspath(__file__)) if '__file__' in globals() else '.'
df = pd.read_csv(os.path.normpath(os.path.join(script_dir, "..", "data", "kepler_cleaned.csv")))

# Set plotting style
sns.set_theme(style="whitegrid")

# Plot 1: Planetary Radius vs Target
plt.figure(figsize=(10, 6))
sns.boxplot(x="target", y="koi_prad", data=df, palette="Set2")
plt.yscale("log")  # Log scale since planetary sizes vary wildly
plt.title("Planetary Radius Distribution by Target (0 = False Positive, 1 = Exoplanet)")
plt.xlabel("Target Class")
plt.ylabel("Planetary Radius (Earth Radii - Log Scale)")
plt.show()
```

```python
# Plot 2: Correlation Matrix for Top Features
plt.figure(figsize=(12, 10))
correlation_matrix = df[['target', 'koi_period', 'koi_duration', 'koi_depth', 'koi_prad', 'koi_steff', 'koi_model_snr']].corr()

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Key Kepler Features")
plt.show()
```
